# NLP vocabulary

Make a list of useful words from a paragraph.


```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

## Try a paragraph

Change `text` to use your own words.


```python
text = "Dogs and cats are playing in a bright garden. The dogs chase a red ball."
```

## Build the vocabulary

Keep nouns, adjectives, and words ending in `-ing`. Skip common words and duplicates.


```python
def make_vocabulary(text):
    # Get common words and split the text into words.
    common_words = set(stopwords.words("english"))
    tagged_words = nltk.pos_tag(nltk.word_tokenize(text.lower()))
    lemmatizer = WordNetLemmatizer()
    words = set()

    for word, tag in tagged_words:
        if not word.isalpha():
            continue

        # Change each useful word to its base form.
        if tag.startswith("NN"):
            base = lemmatizer.lemmatize(word, "n")
        elif tag.startswith("JJ"):
            base = lemmatizer.lemmatize(word, "a")
        elif tag == "VBG":
            base = lemmatizer.lemmatize(word, "v")
        else:
            continue

        if base not in common_words:
            words.add(base)

    return sorted(words)


vocabulary = make_vocabulary(text)
print(vocabulary)
```
